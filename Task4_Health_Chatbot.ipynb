```python
# Task 4: Health Query Chatbot - Final Working Version
# DevelopersHub Corporation - AI/ML Internship

# Knowledge base for health questions
health_knowledge = {
    "sore throat": "Sore throat is commonly caused by viral infections (like cold or flu), bacterial infections (strep throat), allergies, or dry air.",
    "paracetamol": "Paracetamol (acetaminophen) is safe for children when given at the correct dose based on age/weight. Always use children's formula and follow package instructions.",
    "fever": "To reduce fever: rest, drink plenty of fluids, use cool compresses, and take fever-reducing medication like paracetamol or ibuprofen as directed.",
}

def health_chatbot(question):
    """Answer health questions with safety disclaimer"""
    question_lower = question.lower()
    
    if "sore throat" in question_lower:
        answer = health_knowledge["sore throat"]
    elif "paracetamol" in question_lower or "acetaminophen" in question_lower:
        answer = health_knowledge["paracetamol"]
    elif "fever" in question_lower:
        answer = health_knowledge["fever"]
    else:
        answer = "I can provide general information about common health topics like sore throat, fever, and paracetamol safety."
    
    return f"{answer}\n\n⚠️ DISCLAIMER: This is general information only. Consult a doctor for medical advice."

# Test the chatbot with required questions
print("="*60)
print("HEALTH QUERY CHATBOT - TASK 4")
print("="*60)

test_questions = [
    "What causes a sore throat?",
    "Is paracetamol safe for children?",
    "How to reduce fever?"
]

for question in test_questions:
    print(f"\n❓ {question}")
    print(f"🤖 {health_chatbot(question)}")
    print("-"*50)

print("\n✅ Chatbot ready for use.")
```
